```python
#task 3: data cleansing and transformation

#Ben 
#Transforming str variables to in64 variables
def clean_data(data):
    clean = data.copy(deep=True)
    clean["sex"] = clean["sex"].replace({"M": 0, "F": 1, "m": 0, "f": 1})  # Convert sex to numerical values
    clean["region"] = clean["region"].replace({"North": 0, "East": 1, "South": 2, "West": 3, "Central": 4})  # Convert region to numerical values
    clean["urban_rural"] = clean["urban_rural"].replace({"Rural": 0, "Suburban": 1, "Urban": 2})  # Convert urban/rural to numerical values
    clean["education"] = clean["education"].replace({"No Hs": 0, "HS": 1, "Some College": 2, "Bachelors": 3, "Masters": 4, "Doctorate": 5})  # Convert education to numerical values
    clean["marital_status"] = clean["marital_status"].replace({"Single": 0, "Divorced": 1, "Widowed": 2, "Married": 3})  # Convert marital status to numerical values
    clean["employment_status"] = clean["employment_status"].replace({"Unemployed": 0, "Retired": 1, "Employed": 2, "Self-employed": 3})  # Convert employment status to numerical values
    clean["smoker"] = clean["smoker"].replace({"Never": 0, "Former": 1, "Current": 2})  #convert smoker to numerical values
    clean["alcohol_freq"] = clean["alcohol_freq"].replace({"None": 0, "Occasionally": 1, "Weekly": 2, "Daily": 3})  # Convert alcohol frequency to numerical values
    return clean
```
